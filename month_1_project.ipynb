```python
"""
chunking_harness.py
=====================
Runs all 7 chunking strategies against one or more documents (PDF/TXT/MD/HTML)
at multiple target chunk sizes, and produces a comparison table of structural
quality metrics.

Metrics per (document, method, chunk size):
    chunk_count            - how many chunks were produced
    avg_len / median_len   - chunk length in characters
    min_len / max_len
    std_len                - length variance (tighter is usually "cleaner")
    pct_mid_sentence_cut   - % of chunks NOT ending on sentence punctuation
                              (a structural-quality proxy — not a retrieval
                              quality measure; see note at bottom of script)
    elapsed_sec            - processing time

Note: markdown_aware only runs on .md files, html_aware only on .html/.htm
files (they need that structure to split on). The other 5 strategies run on
any text-bearing file (.pdf, .txt, .md).

Usage:
    python chunking_harness.py doc1.pdf doc2.md doc3.html
    python chunking_harness.py --dir ./my_pdfs
    python chunking_harness.py --dir ./my_pdfs --sizes 256 512 --out results.csv

Install:
    pip install langchain-text-splitters pypdf pandas sentence-transformers tiktoken
    (tiktoken and sentence-transformers are optional — the script degrades
    gracefully without them, see the try/except blocks below)
"""

import re
import time
from pathlib import Path

import pandas as pd

from langchain_text_splitters import (
    CharacterTextSplitter,
    RecursiveCharacterTextSplitter,
    MarkdownHeaderTextSplitter,
    HTMLHeaderTextSplitter,
)

# ---------------------------------------------------------------------------
# Optional dependencies — script still runs (with reduced coverage) without them
# ---------------------------------------------------------------------------
try:
    import tiktoken
    _ENC = tiktoken.get_encoding("cl100k_base")

    def count_tokens(text):
        return len(_ENC.encode(text))
except ImportError:
    def count_tokens(text):
        return len(text.split())  # rough word-count fallback


def tokens_to_chars(n_tokens):
    """~4 chars/token is a standard rough ratio for English prose."""
    return int(n_tokens * 4)


try:
    from sentence_transformers import SentenceTransformer
    import numpy as np

    _MODEL = None

    def get_model():
        global _MODEL
        if _MODEL is None:
            _MODEL = SentenceTransformer("all-MiniLM-L6-v2")
        return _MODEL

    SEMANTIC_AVAILABLE = True
except ImportError:
    SEMANTIC_AVAILABLE = False

try:
    from pypdf import PdfReader

    PDF_AVAILABLE = True
except ImportError:
    PDF_AVAILABLE = False


# ---------------------------------------------------------------------------
# Text extraction
# ---------------------------------------------------------------------------

def extract_text(path: Path) -> list[dict]:
    """Returns a list of {"page": int, "text": str} — one entry per PDF page,
    or a single entry (page=1) for non-paginated formats."""
    suffix = path.suffix.lower()
    if suffix == ".pdf":
        if not PDF_AVAILABLE:
            raise RuntimeError("pip install pypdf to read PDFs")
        reader = PdfReader(str(path))
        return [
            {"page": i + 1, "text": (p.extract_text() or "").strip()}
            for i, p in enumerate(reader.pages)
        ]
    if suffix in (".htm", ".html"):
        from bs4 import BeautifulSoup
        raw = path.read_text(encoding="utf-8", errors="ignore")
        soup = BeautifulSoup(raw, "html.parser")
        for tag in soup(["script", "style"]):
            tag.decompose()
        return [{"page": 1, "text": soup.get_text(separator="\n")}]
    return [{"page": 1, "text": path.read_text(encoding="utf-8", errors="ignore")}]



# ---------------------------------------------------------------------------
# The 7 chunking strategies — each returns a flat list of chunk strings
# ---------------------------------------------------------------------------

def chunk_fixed(text, chunk_size_chars, overlap_chars):
    splitter = CharacterTextSplitter(
        separator="\n", chunk_size=chunk_size_chars, chunk_overlap=overlap_chars
    )
    return splitter.split_text(text)


def chunk_recursive(text, chunk_size_chars, overlap_chars):
    splitter = RecursiveCharacterTextSplitter(
        separators=["\n\n", "\n", ". ", " ", ""],
        chunk_size=chunk_size_chars, chunk_overlap=overlap_chars
    )
    return splitter.split_text(text)


def chunk_sliding_window(text, chunk_size_chars, overlap_chars):
    step = max(chunk_size_chars - overlap_chars, 1)
    end = max(len(text) - chunk_size_chars + 1, 1)
    return [text[i:i + chunk_size_chars] for i in range(0, end, step)]

MIN_CHUNK_CHARS = 40  # drop/merge fragments shorter than this

def clean_text(text):
    """Strip PDF TOC/dot-leader junk and bare page numbers before splitting."""
    text = re.sub(r"\.{3,}", " ", text)          # dot leaders: "..... 10"
    text = re.sub(r"^\s*\d+\s*$", "", text, flags=re.MULTILINE)  # bare page numbers
    text = re.sub(r"[ \t]+", " ", text)
    return text

def chunk_semantic(text, chunk_size_chars, overlap_chars, similarity_threshold=0.5):
    if not SEMANTIC_AVAILABLE:
        return []
    text = clean_text(text)
    sentences = [s.strip() for s in re.split(r"(?<=[.!?])\s+", text.strip()) if s.strip()]
    if len(sentences) < 2:
        return sentences

    model = get_model()
    sentence_embeddings = model.encode(sentences)

    chunks, current = [], [sentences[0]]
    for i in range(1, len(sentences)):
        a, b = sentence_embeddings[i - 1], sentence_embeddings[i]
        sim = float(np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b) + 1e-9))
        candidate_len = sum(len(s) for s in current) + len(sentences[i])
        if sim < similarity_threshold or candidate_len > chunk_size_chars:
            if sum(len(s) for s in current) < MIN_CHUNK_CHARS:
                current.append(sentences[i])
                continue
            chunks.append(" ".join(current))
            current = [sentences[i]]
        else:
            current.append(sentences[i])
    if current:
        chunks.append(" ".join(current))

    chunks = [c for c in chunks if len(c) >= MIN_CHUNK_CHARS]

    return chunks


def chunk_parent_child(text, chunk_size_chars, overlap_chars):
    # Compared here on the child (retrieval) unit, since that's what gets
    # matched against queries — the parent is only relevant once you wire
    # this into actual retrieval, not for structural comparison.
    parent_splitter = RecursiveCharacterTextSplitter(
        separators=["\n\n", "\n", ". ", " ", ""],
        chunk_size=chunk_size_chars * 4, chunk_overlap=overlap_chars
    )
    child_splitter = RecursiveCharacterTextSplitter(
        separators=["\n\n", "\n", ". ", " ", ""],
        chunk_size=chunk_size_chars, chunk_overlap=max(overlap_chars // 2, 0)
    )
    children = []
    for parent_text in parent_splitter.split_text(text):
        children.extend(child_splitter.split_text(parent_text))
    return children


def chunk_markdown_aware(text, chunk_size_chars, overlap_chars):
    headers = [("#", "Header 1"), ("##", "Header 2"), ("###", "Header 3")]
    header_splitter = MarkdownHeaderTextSplitter(headers_to_split_on=headers, strip_headers=False)
    sections = header_splitter.split_text(text)
    recursive = RecursiveCharacterTextSplitter(separators=["\n\n", "\n", ". ", " ", ""], chunk_size=chunk_size_chars, chunk_overlap=overlap_chars)
    out = []
    for s in sections:
        if len(s.page_content) <= chunk_size_chars:
            out.append(s.page_content)
        else:
            out.extend(recursive.split_text(s.page_content))
    return out


def chunk_html_aware(text, chunk_size_chars, overlap_chars):
    headers = [("h1", "Header 1"), ("h2", "Header 2"), ("h3", "Header 3")]
    splitter = HTMLHeaderTextSplitter(headers_to_split_on=headers)
    sections = splitter.split_text(text)
    recursive = RecursiveCharacterTextSplitter(separators=["\n\n", "\n", ". ", " ", ""], chunk_size=chunk_size_chars, chunk_overlap=overlap_chars)
    out = []
    for s in sections:
        if len(s.page_content) <= chunk_size_chars:
            out.append(s.page_content)
        else:
            out.extend(recursive.split_text(s.page_content))
    return out


METHODS = {
    # "fixed":         {"fn": chunk_fixed,         "applies_to": {".pdf", ".txt", ".md"}},
    "recursive":     {"fn": chunk_recursive,     "applies_to": {".pdf", ".txt", ".md"}},
    "sliding_window":{"fn": chunk_sliding_window,"applies_to": {".pdf", ".txt", ".md"}},
    "semantic":      {"fn": chunk_semantic,      "applies_to": {".pdf", ".txt", ".md"}},
    "parent_child":  {"fn": chunk_parent_child,  "applies_to": {".pdf", ".txt", ".md"}},
    "markdown_aware":{"fn": chunk_markdown_aware,"applies_to": {".md"}},
    # "html_aware":    {"fn": chunk_html_aware,    "applies_to": {".html", ".htm"}},
}


# ---------------------------------------------------------------------------
# Metrics
# ---------------------------------------------------------------------------

_SENTENCE_END_RE = re.compile(r'[.!?]["\')\]]?\s*$')


def is_mid_sentence_cut(chunk: str) -> bool:
    chunk = chunk.strip()
    if not chunk:
        return False
    return not bool(_SENTENCE_END_RE.search(chunk))


def compute_metrics(chunks, elapsed_sec):
    if not chunks:
        return dict(chunk_count=0, avg_len=0, median_len=0, min_len=0, max_len=0,
                     std_len=0, pct_mid_sentence_cut=0, elapsed_sec=round(elapsed_sec, 4))
    lengths = pd.Series([len(c) for c in chunks])
    mid_cuts = sum(is_mid_sentence_cut(c) for c in chunks)
    return dict(
        chunk_count=len(chunks),
        avg_len=round(lengths.mean(), 1),
        median_len=round(lengths.median(), 1),
        min_len=int(lengths.min()),
        max_len=int(lengths.max()),
        std_len=round(lengths.std() or 0, 1),
        pct_mid_sentence_cut=round(100 * mid_cuts / len(chunks), 1),
        elapsed_sec=round(elapsed_sec, 4),
    )


# ---------------------------------------------------------------------------
# Harness
# ---------------------------------------------------------------------------

def run_harness(paths, chunk_size_tokens_list=(256, 512), overlap_pct=15):
    rows = []
    all_chunks = []  # flat list of chunk records for the JSON export
    chunk_counter = 0

    for path in paths:
        path = Path(path)
        print(f"Processing {path.name} ...")
        pages = extract_text(path)
        suffix = path.suffix.lower()

        for target_tokens in chunk_size_tokens_list:
            chunk_size_chars = tokens_to_chars(target_tokens)
            overlap_chars = chunk_size_chars * overlap_pct // 100

            for method_name, spec in METHODS.items():
                if suffix not in spec["applies_to"]:
                    continue

                start = time.perf_counter()
                method_chunks = []  # (chunk_text, page) pairs for this method/size
                try:
                    for page_entry in pages:
                        page_chunks = spec["fn"](page_entry["text"], chunk_size_chars, overlap_chars)
                        # chunk_semantic can return (chunks, embeddings) — unwrap if so
                        if isinstance(page_chunks, tuple):
                            page_chunks = page_chunks[0]
                        method_chunks.extend((c, page_entry["page"]) for c in page_chunks)
                except Exception as e:
                    print(f"  [{method_name} @ {target_tokens}tok] FAILED: {e}")
                    continue
                elapsed = time.perf_counter() - start

                just_text = [c for c, _ in method_chunks]
                metrics = compute_metrics(just_text, elapsed)
                metrics.update(document=path.name, method=method_name, target_tokens=target_tokens)
                rows.append(metrics)

                for chunk_text, page_num in method_chunks:
                    all_chunks.append({
                        "chunk_id": f"chunk_{chunk_counter}",
                        "text": chunk_text,
                        "source": path.name,
                        "page": page_num,
                        "method": method_name,
                        "target_tokens": target_tokens,
                    })
                    chunk_counter += 1

    df = pd.DataFrame(rows)
    cols = ["document", "method", "target_tokens", "chunk_count", "avg_len",
            "median_len", "min_len", "max_len", "std_len",
            "pct_mid_sentence_cut", "elapsed_sec"]
    df = df[cols] if not df.empty else df
    return df, all_chunks


def print_summary(df: pd.DataFrame):
    if df.empty:
        print("No results — check your inputs.")
        return

    print("\n" + "=" * 100)
    print("FULL RESULTS (per document x method x chunk size)")
    print("=" * 100)
    print(df.to_string(index=False))

    print("\n" + "=" * 100)
    print("AVERAGED ACROSS ALL DOCUMENTS (per method x chunk size)")
    print("=" * 100)
    summary = (
        df.groupby(["method", "target_tokens"])
        .agg(
            avg_chunk_count=("chunk_count", "mean"),
            avg_len=("avg_len", "mean"),
            avg_std_len=("std_len", "mean"),
            avg_pct_mid_sentence_cut=("pct_mid_sentence_cut", "mean"),
            avg_elapsed_sec=("elapsed_sec", "mean"),
        )
        .round(2)
        .reset_index()
        .sort_values(["target_tokens", "avg_pct_mid_sentence_cut"])
    )
    print(summary.to_string(index=False))
    print(
        "\nReminder: pct_mid_sentence_cut is a STRUCTURAL proxy, not a retrieval-quality\n"
        "score. A method can win here and still lose on IoU/recall if it groups the\n"
        "wrong sentences together. Use this table to catch broken/degenerate splitters\n"
        "fast, then run the winners through chunking_evaluation (IoU/recall) before\n"
        "picking a default."
    )


# ---------------------------------------------------------------------------
# CONFIG — edit directly and re-run the cell/script. No argparse/CLI args,
# so this works the same whether you `python chunking_harness.py` locally
# or paste the whole file into a Colab cell and hit Run (Colab's own kernel
# args would break argparse, so we don't use it).
# ---------------------------------------------------------------------------

DOCS = [
    # add your own paths here, e.g. "/content/sample_docs/my_report.pdf"
]
CHUNK_SIZES_TOKENS = [512]
OVERLAP_PCT = 15
OUTPUT_CSV = "comparison_results.csv"

# Optional: point this at a folder instead of listing files individually.
# Auto-defaults to "sample_docs" (or "/content/sample_docs" on Colab) if that
# folder exists, so this works out of the box right after the download cell.
DOCS_DIR = None  # override e.g. "/content/sample_docs"

if DOCS_DIR is None:
    for _candidate in ("/content/sample_docs", "sample_docs"):
        if Path(_candidate).is_dir():
            DOCS_DIR = _candidate
            break


def _resolve_docs():
    paths = list(DOCS)
    if DOCS_DIR:
        supported = {".pdf", ".txt", ".md", ".html", ".htm"}
        paths += sorted(str(p) for p in Path(DOCS_DIR).iterdir() if p.suffix.lower() in supported)
    return paths


if __name__ == "__main__":
    import json

    input_paths = _resolve_docs()
    if not input_paths:
        raise SystemExit(
            "No documents found. Either:\n"
            "  1) Run the download cell first so ./sample_docs exists (auto-detected), or\n"
            "  2) Edit the DOCS list or DOCS_DIR near the bottom of this script, then re-run this cell."
        )
    results_df, all_chunks = run_harness(input_paths, chunk_size_tokens_list=CHUNK_SIZES_TOKENS, overlap_pct=OVERLAP_PCT)
    print_summary(results_df)

    summary_df = (
        results_df.groupby(["method", "target_tokens"])
        .agg(
            avg_chunk_count=("chunk_count", "mean"),
            avg_len=("avg_len", "mean"),
            avg_std_len=("std_len", "mean"),
            avg_pct_mid_sentence_cut=("pct_mid_sentence_cut", "mean"),
            avg_elapsed_sec=("elapsed_sec", "mean"),
        )
        .round(2)
        .reset_index()
    )

    excel_path = OUTPUT_CSV.replace(".csv", ".xlsx")
    with pd.ExcelWriter(excel_path) as writer:
        results_df.to_excel(writer, sheet_name="full_results", index=False)
        summary_df.to_excel(writer, sheet_name="summary", index=False)
    print(f"\nSaved summary workbook to {excel_path}")

    chunks_path = OUTPUT_CSV.replace(".csv", "_chunks.json")
    with open(chunks_path, "w", encoding="utf-8") as f:
        json.dump(all_chunks, f, ensure_ascii=False, indent=2)
    print(f"Saved {len(all_chunks)} chunks to {chunks_path}")
```

```python
def filter_chunks(df, method=None, target_tokens=None, source=None):
    """Filter chunks_df by any combination of metadata fields before indexing."""
    out = df
    if method is not None:
        out = out[out["method"] == method]
    if target_tokens is not None:
        out = out[out["target_tokens"] == target_tokens]
    if source is not None:
        out = out[out["source"] == source]
    return out.reset_index(drop=True)

# Set this after reading your Excel summary sheet and picking the winner
# (lowest pct_mid_sentence_cut / best std_len tradeoff for your data).
CHOSEN_METHOD = "semantic"
CHOSEN_TOKENS = 512

chunks_df = filter_chunks(chunks_df, method=CHOSEN_METHOD, target_tokens=CHOSEN_TOKENS)
chunks_df = chunks_df.reset_index(drop=True)

corpus = chunks_df["text"].tolist()
source = chunks_df["doc_id"].tolist()

print(f"Indexing {len(corpus)} chunks from method={CHOSEN_METHOD}, size={CHOSEN_TOKENS}")
```

```python
def get_bm25(corpus, stop_words=[]):
    tokens = bm25s.tokenize(corpus, stopwords=stop_words)
    bm25_retriever = bm25s.BM25()
    bm25_retriever.index(tokens, corpus)
    return bm25_retriever


bm25_index = get_bm25(corpus=corpus, stop_words="en")
```

```python
def search_dense(collection, query: str, k: int = 10):
    result = collection.query(query_texts=[query], n_results=k)
    ids = [int(doc_id.split("_")[1]) for doc_id in result["ids"][0]]
    distances = result["distances"][0]
    return list(zip(ids, distances)), result


def search_bm25(query: str, k: int = 10):
    query_tokens = bm25s.tokenize([query], stopwords="en")
    indices, scores = bm25_index.retrieve(query_tokens, k=k)
    return indices, scores


K_RRF = 60


def reciprocal_rank_fusion(rankings: list, k: int = K_RRF):
    scores: dict = defaultdict(float)
    for ranking in rankings:
        for rank, doc_id in enumerate(ranking, start=1):
            scores[doc_id] += 1.0 / (k + rank)
    return sorted(scores.items(), key=lambda x: -x[1])


def search_hybrid(collection, query: str, k: int = 10, candidate_k: int = 10):
    bm25_raw_indices, _ = search_bm25(query, k=candidate_k)
    bm25_doc_ids = [str(doc_id) for doc_id in bm25_raw_indices[0]] if bm25_raw_indices.size > 0 else []

    dense_raw_results, _ = search_dense(collection, query, k=candidate_k)
    dense_doc_ids = [str(doc_id) for doc_id, _ in dense_raw_results]

    fused_results = reciprocal_rank_fusion([bm25_doc_ids, dense_doc_ids])
    return fused_results[:k], dense_doc_ids, bm25_doc_ids


def normalize_retrieval(method_name, query, k_candidates=20):
    if method_name == "dense":
        raw, _ = search_dense(collection, query, k=k_candidates)
        return [(int(pid), score) for pid, score in raw]
    if method_name == "bm25":
        indices, scores = search_bm25(query, k=k_candidates)
        return [(int(pid), float(s)) for pid, s in zip(indices[0], scores[0])]
    if method_name == "hybrid":
        fused, _, _ = search_hybrid(collection, query, k=k_candidates, candidate_k=k_candidates)
        return [(int(pid), score) for pid, score in fused]
    raise ValueError(f"unknown method: {method_name}")


def position_to_doc_id(position: int) -> str:
    return source[position]
```

```python
_RERANKER = None


def get_reranker():
    global _RERANKER
    if _RERANKER is None:
        _RERANKER = FlagReranker("BAAI/bge-reranker-v2-m3", use_fp16=True)
    return _RERANKER


def reranking(query, chunks, k=None):
    """chunks: list of [text, id]. Returns [[score, id], ...] sorted desc."""
    reranker = get_reranker()
    pairs = [[query, text] for text, _id in chunks]
    scores = reranker.compute_score(pairs)
    if isinstance(scores, float):
        scores = [scores]
    result = sorted(
        ([score, _id] for score, (_, _id) in zip(scores, chunks)),
        key=lambda x: x[0],
        reverse=True,
    )
    return result[:k] if k else result


def rerank_retrieval(method_name, query, k=10, k_candidates=20, rerank_fn=reranking):
    before = normalize_retrieval(method_name, query, k_candidates=k_candidates)
    chunk_pairs = [[corpus[pos], pos] for pos, _ in before]
    reranked = rerank_fn(query, chunk_pairs)
    return {
        "before_ids": [pos for pos, _ in before][:k],
        "before_scores": [s for _, s in before][:k],
        "after_ids": [pos for _, pos in reranked][:k],
        "after_scores": [s for s, _ in reranked][:k],
    }
```

```python
# id -> metadata lookup, built once, used to filter retrieval results before generation
position_to_meta = {
    j: {
        "doc_id": source[j],
        "page": chunks_df["page"].iloc[j],
        "method": chunks_df["method"].iloc[j],
    }
    for j in range(len(corpus))
}

def filter_positions_by_metadata(positions, source_filter=None, page_range=None):
    """Drop retrieved chunk positions that don't match the given metadata constraints."""
    filtered = []
    for pos in positions:
        meta = position_to_meta.get(pos, {})
        if source_filter is not None and meta.get("doc_id") != source_filter:
            continue
        if page_range is not None:
            page = meta.get("page")
            if page is None or not (page_range[0] <= page <= page_range[1]):
                continue
        filtered.append(pos)
    return filtered


def retrieve_contexts_config(query: str, method: str, use_rerank: bool, top_k: int = 5,
                              k_candidates: int = 20, source_filter=None, page_range=None):
    """Same as before, with an added metadata filter step before contexts are built."""
    t0 = time.perf_counter()

    if use_rerank:
        r = rerank_retrieval(method, query, k=k_candidates, k_candidates=k_candidates, rerank_fn=reranking)
        positions = r["after_ids"]
    else:
        raw = normalize_retrieval(method, query, k_candidates=k_candidates)
        positions = [pos for pos, _ in raw]

    positions = filter_positions_by_metadata(positions, source_filter=source_filter, page_range=page_range)
    positions = positions[:top_k]

    latency = time.perf_counter() - t0
    contexts = [corpus[pos] for pos in positions]
    return contexts, latency
```

```python
def generate_answer(query: str, contexts: list) -> str:
    context_text = "\n\n".join(contexts)
    prompt = f"""Answer the question directly and concisely, using ONLY the context below.
Do not comment on whether the context is complete or sufficient — just answer with whatever relevant information is present.
Only if the context contains genuinely nothing relevant to the question, respond with exactly: "I don't know based on the provided context."

Context:
{context_text}

Question: {query}

Answer:"""
    response = gemini_client.models.generate_content(
        model="gemini-2.5-flash-lite",
        contents=prompt,
    )
    return response.text.strip()
```

```python
CONFIGS = [
    ("bm25", False), ("bm25", True),
    ("dense", False), ("dense", True),
    ("hybrid", False), ("hybrid", True),
]

def build_dataset_for_config(method, use_rerank, rows):
    data = {"question": [], "answer": [], "contexts": [], "ground_truth": []}
    latencies = []
    for qid, query_text, gt in rows:  # your merged 40-60 row set
        contexts, lat = retrieve_contexts_config(query_text, method, use_rerank)
        answer = generate_answer(query_text, contexts)
        data["question"].append(query_text)
        data["answer"].append(answer)
        data["contexts"].append(contexts)
        data["ground_truth"].append(gt)
        latencies.append(lat)
    return Dataset.from_dict(data), sum(latencies) / len(latencies)
```
